Write a function that takes three numbers
as arguments and returns the largest
number


```python
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

print("Largest number is:", largest(num1, num2, num3))
```

Write a function that takes a string and
returns the number of vowels (a, e, i, o, u)
present in it

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for ch in text:
        if ch in vowels:
            count += 1
    return count

string = input("Enter a string: ")

print("Number of vowels:", count_vowels(string))
```

  ENUMERATE:In Python, the enumerate() function used to iterate over a sequence (such as a list or tuple) and return a
sequence of tuples containing the index and value of each element in the original sequence.

enumerate(iterable, start=0)
iterable → the collection you want to loop through.
start → the starting index (default is 0).

```python
values=[1,2,3,4,5,6,7]
```

```python
for i in enumerate(values):
  print(i)
```

```python
v=["ram","shyam","sita","gita"]
```

```python
for i, value in enumerate(v,start=0):
  print(i,value)
```

ZIP():The zip() function in Python is used to combine two or more iterables (e.g., lists or tuples) into a single iterable that
returns tuples containing the corresponding elements from each input iterable.
The general syntax for zip() is:
zip(iterable1, iterable2, ...)
where iterable1, iterable2, etc. are the iterables to be zipped. The zip() function returns an iterator of tuples, where the ith tuple contains the i-th element from each of the input iterables. The resulting iterator has as many elements as the
shortest input iterable

```python
lst=["name","gender","dob","salary"]
lst2=["abcd","male","2045-05-05","10000"]
```

```python
for i, (k, v) in enumerate(zip(lst, lst2), start=1):
    print(f"{i}. {k} : {v}")
```

```python
print(list(zip(lst,lst2)))
```

map():The map() function in Python is a built-in function that allows you to apply a given function to every item in an iterable (like a list or tuple). Essentially, it takes two main arguments: a function and an iterable. The function is applied to each element of the iterable, and map() returns a map object, which you can convert to a list, tuple, or another type.

In short, instead of writing a loop yourself, map() does it for you, applying the function to each element automatically. It’s a very clean way to transform data!

The map() function in Python applies a given function to each item of an iterable (e.g., a list or a tuple) and
returns a new iterable with the transformed items.
The general syntax for map() is:
map(function, iterable)
where function is the function to be applied to each item in the iterable, and iterable is the object to be mapped.
The map() function returns a map object, which is an iterable of the transformed items. To get a list of the
transformed items, you can convert the map object to a list using the list() function

```python
def square(x):
    return x**2

numbers = [1, 2, 3, 4, 5]
squares = list(map(square, numbers))
print(squares)
```
