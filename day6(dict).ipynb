del  → Delete only

pop  → Pop out and give me the value

```python
data = {
    "name": "Hari",
    "age": 20
}

del data["age"]        # just delete

```

```python
data = {
    "name": "Hari",
    "age": 20
}



age = data.pop("age")  # delete and return value
```

```python
data
```

del removes a key-value pair from a dictionary but does not return the removed value.

pop() removes a key-value pair and returns the removed value. It can also take a default value to avoid errors if the key is missing.

```python

dict1={1:2,3:4}
dict2={5:6,7:8}
dict1.update(dict2)
print(dict1)
```
