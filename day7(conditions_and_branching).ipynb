```python
age = int(input("Enter your age: "))
is_citizen = input("Are you a citizen? (yes/no): ").lower()
has_citizenship_card = input("Do you have your Citizenship Card? (yes/no): ").lower()
has_voter_card = input("Do you have your Voter ID Card? (yes/no): ").lower()

if age < 18:
    print("Status: NOT ELIGIBLE")
    print("Reason: You must be 18 or older.")
elif is_citizen != "yes":
    print("Status: NOT ELIGIBLE")
    print("Reason: You must be a citizen.")
elif has_citizenship_card != "yes" and has_voter_card != "yes":
    print("Status: NOT ELIGIBLE")
    print("Reason: You must bring either your Citizenship Card or Voter Card.")
else:
    print("Status: ELIGIBLE TO VOTE")
    print("Reason: All requirements met.")
```

```python
score = float(input("Enter your SEE percentage (0-100): "))

if score < 0 or score > 100:
    print("Invalid percentage! Please enter a value between 0 and 100.")
elif score >= 90:
    print("Grade: A+ | CGPA: 4.0 | Remarks: OUTSTANDING")
elif score >= 80:
    print("Grade: A  | CGPA: 3.6 | Remarks: EXCELLENT")
elif score >= 70:
    print("Grade: B+ | CGPA: 3.2 | Remarks: VERY GOOD")
elif score >= 60:
    print("Grade: B  | CGPA: 2.8 | Remarks: GOOD")
elif score >= 50:
    print("Grade: C+ | CGPA: 2.4 | Remarks: SATISFACTORY")
elif score >= 40:
    print("Grade: C  | CGPA: 2.0 | Remarks: ACCEPTABLE")
elif score >= 30:
    print("Grade: D  | CGPA: 1.6 | Remarks: BASIC")
else:
    print("Grade: NG | CGPA: -   | Remarks: NOT GRADED")
```

```python
age = int(input("Enter your age: "))
is_membership = input("Do you have a membership? (true/false): ").lower() == "true"

if age >= 40 and is_membership:
    discount = 60
elif age >= 40 and not is_membership:
    discount = 40
else:
    discount = 0

print(f"Your discount is: {discount}%")
```
