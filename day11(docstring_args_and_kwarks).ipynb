reduce():The reduce() function in Python is used to apply a given function to the items of an iterable (e.g., a list or a tuple)
in a cumulative way, that is, it applies the function to the first two elements, then to the result and the next
element, and so on, until all the items have been processed. The general syntax for reduce() is:
reduce(function, iterable[, initializer])
where function is the function to be applied cumulatively to the items of the iterable, and initializer is an
optional initial value to be used as the starting point of the cumulative computation. If initializer is not provided,
the first two elements of the iterable are used as the initial values.
Here's an example that demonstrates how to use reduce() to compute the product of all the elements in a list:


```python
from functools import reduce

def mux(a, b):
    return (a * b)
lst1 = [1, 2, 3, 4, 5]
resdata = reduce(mux, lst1)
resdata
```

```python
from functools import reduce

def maximum(x, y):
    if x > y:
        return x
    else:
        return y

print(reduce(maximum, [4, 9, 2, 8, 1]))
```

*args and **kwargs
*args → arguments (many values)
**kwargs → keyword arguments (key=value)

```python
def data(*args):
    return args

print(data("welcome", "to", "skillshikshya"))
```

```python
def data(**kargs):
    return kargs

print(data(name='dipsan',age=23))
```

```python
def data3(n,*args,**kwargs):
  print(n)
  print(args)
  print(kwargs)
data3(3,2,3,4,5,s=1,s1=4)
```

```python
def pizza_order(n, *args, **kwargs):
    print("\n PIZZA ORDER SUMMARY")
    print("Number of pizzas:", n)

    print("\nPizza Types:")
    for i, pizza in enumerate(args, 1):
        print(f"{i}. {pizza}")

    print("\nCustomer Info:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")


n = int(input("Enter number of pizzas: "))

lst_1 = []

for i in range(n):
    pizza = input(f"Enter pizza type {i+1}: ")
    lst_1.append(pizza)

name = input("Enter your name: ")
address = input("Enter your address: ")

pizza_order(n, *lst_1, name=name, address=address)
```
