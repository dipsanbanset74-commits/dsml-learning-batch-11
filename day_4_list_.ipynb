Slicing in list

```python
num=[10,20,30,40,50,60]
print(num[1:4]) #elements from index 1 to 3
print(num[:3]) #first three element
print(num[3:]) #from index 3 to end
print(num[::2]) #every second element
print(num[::-1])#reverse list
print(num[:])#creates a shallow copy of entire list
```

List as Queues

```python
from collections import deque
#deque means Double Ended Queue
queue = deque(["DIPSAN", "PRANJALI", "TULSI"])
queue.append("PRANJALI")
queue.append("DIPSAN")
queue.popleft()
queue.popleft()
queue
```

deep copy and shallow copy
image.png



Simple Real-Life Example

Suppose:

Original notebook has attached pages.
Shallow Copy

You copy only notebook cover, but pages are shared.

Deep Copy

You copy both cover and all pages separately.

```python
lst=[1,2,3,4]
lst1=lst
print(lst)
print(lst1)
```

```python
lst2=lst1.copy()
lst
```

```python
import copy

data1 = [[1,2], [3,4]]

data2 = copy.deepcopy(data1)

data2[0][0] = 100

print(data1)
print(data2)
```
