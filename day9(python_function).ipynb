making calculator

```python
def calculator(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b != 0:
            return a / b
        else:
            return "Error: Division by zero"
    else:
        return "Invalid operator"

# User input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

# Function call
result = calculator(num1, num2, operator)

print("Result =", result)
```

```python
def area(length, breadth):
    return length * breadth

# Example usage
l = float(input("Enter length: "))
b = float(input("Enter breadth: "))

print("Area of rectangle =", area(l, b))
```

Write a user-defined function that takes 5 user inputs.
If the user enters an integer, add it to a list
If the user enters a float number, remove the last element
Finally, display the updated list after each operation



```python
def process_inputs():
    lst = []

    for i in range(5):
        value = input(f"Enter value {i+1}: ")


        if '.' in value:
            num = float(value)
            lst.append(num)
            if lst:
                lst.pop()

        else:
            num = int(value)
            lst.append(num)

        print("Updated list:", lst)

process_inputs()
```
