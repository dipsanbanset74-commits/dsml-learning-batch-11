WEBSITE PASSWORD VALIDATION

```python
password = input("Enter password: ")

checks = [
    any(char.isupper() for char in password),  # At least one uppercase
    any(char.islower() for char in password),  # At least one lowercase
    any(char.isdigit() for char in password),  # At least one number
    len(password) >= 8                         # Minimum 8 characters
]

if all(checks):
    print("Valid Password")
else:
    print("Invalid Password")
```
